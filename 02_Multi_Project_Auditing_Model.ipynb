
 Observation: Previously, Revised Budget was mistakenly set lower than Original Budget (indicating budget contraction).
  
 Audit Insight: In public infrastructure projects, revised budgets typically increase due to inflation, material price spikes, and scope expansion (Cost Overrun).

Adjustment: Standardized the dataset where `Revised_Budget_Crore > Original_Budget_Crore` to accurately compute positive Cost Overrun percentages.
