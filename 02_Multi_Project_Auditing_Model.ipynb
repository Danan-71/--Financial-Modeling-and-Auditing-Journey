```python
import pandas as pd

Project_Data = {
    'Project_ID': ['Project A', 'Project B', 'Project C'],
    'Original_Budget_Crore': [800, 300, 500],
    'Prevised_Budget_Crore': [750, 200, 450],
}

df_projects = pd.DataFrame(Project_Data)

df_projects['Cost_Overrun_Pct'] = ((df_projects['Prevised_Budget_Crore'] - df_projects['Original_Budget_Crore']) / df_projects['Original_Budget_Crore']) *100

print("=== ADVANCED AUDIT MODEL ===")
print(df_projects[['Project_ID', 'Cost_Overrun_Pct']])
```
